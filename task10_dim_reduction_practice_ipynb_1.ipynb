# Сокращение размерности

В каждом из заданий:
* Обучите два классификатора (svm и огистическую регрессию на датасете данном ниже) в датасете размерность из 30 признаков
* Попребирайте различное количество итоговых размерностей и постройте графики качества от количества размерностей. Выведите обе модели на одном графике в каждой заадче

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import decomposition
from sklearn import datasets
from sklearn import svm
from sklearn.linear_model import LogisticRegression
np.random.seed(5)
dataset = datasets.load_breast_cancer()
X = dataset.data
y = dataset.target
```

# Forward selection (2pts)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Разделение данных на обучающую и тестовую выборки
X_tr, X_val, y_tr, y_val = train_test_split(X, y, test_size=0.3, random_state=5)

# Получение индексов всех признаков
feature_indices = list(range(X_tr.shape[1]))

# Инициализация списков для хранения результатов
svm_chosen_features = []
lr_chosen_features = []
svm_accuracies = []
lr_accuracies = []
step_numbers = []

def assess_classifier(classifier, selected_features):
    selected = sorted(selected_features)
    X_tr_selected = X_tr[:, selected]
    X_val_selected = X_val[:, selected]

    classifier.fit(X_tr_selected, y_tr)
    y_pred = classifier.predict(X_val_selected)

    return accuracy_score(y_val, y_pred)

for iteration in range(len(feature_indices)):
    best_svm_feature = None
    highest_svm_accuracy = 0

    for feat in feature_indices:
        if feat in svm_chosen_features:
            continue

        potential_features = svm_chosen_features + [feat]
        svm_model = LinearSVC(random_state=123, max_iter=2000)
        accuracy = assess_classifier(svm_model, potential_features)

        if accuracy > highest_svm_accuracy:
            best_svm_feature = feat
            highest_svm_accuracy = accuracy

    if best_svm_feature is not None:
        svm_chosen_features.append(best_svm_feature)

    svm_final_model = LinearSVC(random_state=123, max_iter=2000)
    svm_accuracies.append(assess_classifier(svm_final_model, svm_chosen_features))

    best_lr_feature = None
    highest_lr_accuracy = 0

    for feat in feature_indices:
        if feat in lr_chosen_features:
            continue

        potential_features = lr_chosen_features + [feat]
        lr_model = LogisticRegression(random_state=123, max_iter=1000)
        accuracy = assess_classifier(lr_model, potential_features)

        if accuracy > highest_lr_accuracy:
            best_lr_feature = feat
            highest_lr_accuracy = accuracy

    if best_lr_feature is not None:
        lr_chosen_features.append(best_lr_feature)

    lr_final_model = LogisticRegression(random_state=123, max_iter=1000)
    lr_accuracies.append(assess_classifier(lr_final_model, lr_chosen_features))

    step_numbers.append(iteration + 1)

    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(step_numbers, svm_accuracies, color='blue', linestyle='-', marker='o', label='Метод опорных векторов')
    ax.plot(step_numbers, lr_accuracies, color='red', linestyle='--', marker='s', label='Логистическая регрессия')
    ax.set_xlabel('Количество отобранных признаков', fontsize=12)
    ax.set_ylabel('Точность классификации', fontsize=12)
    ax.set_title(f'Метод последовательного отбора признаков (шаг {iteration+1})', fontsize=14)
    ax.grid(True, alpha=0.3)
    ax.legend(loc='lower right')
    plt.tight_layout()
    plt.show()
```

# Backward elimination (2pts)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X_train_data, X_test_data, y_train_data, y_test_data = train_test_split(X, y, test_size=0.3, random_state=5)

feature_ids = list(range(X_train_data.shape[1]))

iteration_count = []
svm_scores_list = []
lr_scores_list = []

svm_current_set = feature_ids.copy()
lr_current_set = feature_ids.copy()

def evaluate_model_performance(classifier, selected_features):
    sorted_features = sorted(selected_features)
    X_train_reduced = X_train_data[:, sorted_features]
    X_test_reduced = X_test_data[:, sorted_features]
    classifier.fit(X_train_reduced, y_train_data)
    predictions = classifier.predict(X_test_reduced)
    return accuracy_score(y_test_data, predictions)

for step_num in range(len(feature_ids) - 1):
    svm_model = LinearSVC(random_state=123, max_iter=2000)
    current_svm_score = evaluate_model_performance(svm_model, svm_current_set)
    svm_scores_list.append(current_svm_score)

    lr_model = LogisticRegression(random_state=123, max_iter=750)
    current_lr_score = evaluate_model_performance(lr_model, lr_current_set)
    lr_scores_list.append(current_lr_score)

    iteration_count.append(step_num)

    svm_feature_to_remove = None
    highest_svm_score_removed = -1

    for feature in svm_current_set:
        temp_features = [f for f in svm_current_set if f != feature]
        if not temp_features:
            continue
        svm_temp_model = LinearSVC(random_state=123, max_iter=2000)
        temp_score = evaluate_model_performance(svm_temp_model, temp_features)
        if temp_score > highest_svm_score_removed:
            highest_svm_score_removed = temp_score
            svm_feature_to_remove = feature

    lr_feature_to_remove = None
    highest_lr_score_removed = -1

    for feature in lr_current_set:
        temp_features = [f for f in lr_current_set if f != feature]
        if not temp_features:
            continue
        lr_temp_model = LogisticRegression(random_state=123, max_iter=750)
        temp_score = evaluate_model_performance(lr_temp_model, temp_features)
        if temp_score > highest_lr_score_removed:
            highest_lr_score_removed = temp_score
            lr_feature_to_remove = feature

    if svm_feature_to_remove is not None:
        svm_current_set.remove(svm_feature_to_remove)

    if lr_feature_to_remove is not None:
        lr_current_set.remove(lr_feature_to_remove)

plt.figure(figsize=(10, 6))
plt.plot(iteration_count, svm_scores_list, 'b-', linewidth=2.5, label='SVM', marker='o', markersize=6)
plt.plot(iteration_count, lr_scores_list, 'r-', linewidth=2.5, label='Logistic Regression', marker='s', markersize=6)
plt.xlabel('Количество удаленных признаков', fontsize=12)
plt.ylabel('Точность классификации', fontsize=12)
plt.title('Метод обратного исключения признаков', fontsize=14)
plt.grid(True, alpha=0.3, linestyle='--')
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

# PCA (2 pts)


### EM алгоритм

Оставлю это тут просто чтобы пугать людей


$$
\log p(x|\theta) = \int q(z)\log p(x|\theta)dz = \int q(z)\log\dfrac{p(x,z|\theta)}{p(z|x,\theta)}dz = \int q(z)\log\dfrac{p(x,z|\theta)}{q(z)}dz+\int q(z)\log\dfrac{q(z)}{p(z|x,\theta)}dz = \mathcal{L}(q,\theta)+KL(q(z)|p(z|x,\theta))
$$

** E-step **
$$ q(z)^{(n+1)} = p(z|x,\theta^{(n)}) $$
** M-step **
$$ \theta^{(n+1)} = \max\limits_{\theta} \mathcal{L}(q^{(n+1)},\theta) $$

Обучите два классификатора (svm и огистическую регрессию на датасете данном ниже) в датасете размерность из 30 признаков. Попребирайте различное количество итоговых размерностей (до которых будет сокращать PCA) и постройте графики для обоих классификаторов. Сделайте выводы.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import decomposition
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

n_comp_values = []
svm_accuracies = []
lr_accuracies = []

def calculate_model_accuracy(model, X_tr_data, X_te_data, y_tr_data, y_te_data):
    model.fit(X_tr_data, y_tr_data)
    y_pred = model.predict(X_te_data)
    return accuracy_score(y_te_data, y_pred)

for comp_num in range(1, X.shape[1] + 1):
    pca_transformer = decomposition.PCA(n_components=comp_num)

    X_transformed = pca_transformer.fit_transform(X)
    X_tr_split, X_te_split, y_tr_split, y_te_split = train_test_split(
        X_transformed, y, test_size=0.3, random_state=5
    )

    svm_clf = LinearSVC(random_state=69, max_iter=1500)
    lr_clf = LogisticRegression(random_state=69, max_iter=800)

    n_comp_values.append(comp_num)
    svm_accuracies.append(calculate_model_accuracy(svm_clf, X_tr_split, X_te_split, y_tr_split, y_te_split))
    lr_accuracies.append(calculate_model_accuracy(lr_clf, X_tr_split, X_te_split, y_tr_split, y_te_split))

plt.figure(figsize=(10, 6))
plt.plot(n_comp_values, svm_accuracies, 'b-', linewidth=2.5, marker='o', markersize=6, label='Метод опорных векторов')
plt.plot(n_comp_values, lr_accuracies, 'r--', linewidth=2.5, marker='s', markersize=6, label='Логистическая регрессия')
plt.xlabel('Количество главных компонент', fontsize=12)
plt.ylabel('Точность классификации', fontsize=12)
plt.title('Сравнение производительности моделей при PCA', fontsize=14)
plt.grid(True, alpha=0.3, linestyle=':')
plt.legend(fontsize=11, loc='best')
plt.tight_layout()
plt.show()
```

## Выводы по заданию "Сокращение размерности"

### 1. Метод последовательного отбора признаков (Forward Selection):
- Точность обеих моделей (SVM и логистической регрессии) увеличивается с добавлением первых нескольких признаков
- После достижения определенного порога (обычно 5-15 признаков) дальнейшее добавление признаков незначительно влияет на качество моделей или даже может приводить к переобучению
- Разные модели могут иметь разные оптимальные наборы признаков, что демонстрирует разную природу алгоритмов

### 2. Метод обратного исключения признаков (Backward Elimination):
- При удалении малозначимых признаков точность моделей может сохраняться или даже улучшаться
- Обычно можно удалить 10-20 признаков без существенного падения качества классификации
- Уменьшение размерности данным методом часто приводит к более стабильным и интерпретируемым моделям

### 3. Метод главных компонент (PCA):
- Первые несколько главных компонент (обычно 2-5) содержат большую часть информации, необходимой для классификации
- С увеличением числа компонент точность растет, но после определенного порога (обычно 5-10 компонент) прирост становится незначительным
- PCA эффективно снижает размерность, сохраняя максимальную дисперсию данных

### Общие наблюдения:
1. Обе модели демонстрируют схожую динамику изменения точности при изменении размерности данных
2. Часто можно сократить размерность в 2-3 раза (с 30 до 10-15 признаков) без существенной потери точности
3. Каждый метод имеет свои преимущества:
   - **Forward Selection** лучше сохраняет интерпретируемость признаков
   - **Backward Elimination** позволяет удалить наименее информативные признаки
   - **PCA** лучше справляется с мультиколлинеарностью и шумом в данных
4. Оптимальное количество признаков/компонент зависит от конкретной задачи и данных, но обычно находится в диапазоне 5-15

Для данного набора данных рекомендуется использовать PCA с 8-10 компонентами как наиболее сбалансированный подход.
